```python
import numpy as np
from google  import genai
```

```python
client=genai.Client(api_key="............................................")
```

```python
chunks = [
    "Python is a programming language.",
    "Machine learning is a subset of AI.",
    "Deep learning uses neural networks.",
    "Cats are mammals.",
    "Dogs are loyal animals.",
    "The Earth revolves around the Sun.",
    "FastAPI is a modern Python web framework.",
    "Embeddings convert text into vectors.",
    "RAG combines retrieval and generation.",
    "Vector databases store embeddings.",
    "Neural networks learn patterns.",
    "Transformers are powerful NLP models.",
    "OpenCV is used for computer vision.",
    "YOLO is an object detection model.",
    "Pandas is used for data analysis.",
    "NumPy supports numerical computing.",
    "Scikit-learn provides ML algorithms.",
    "LLMs predict the next token.",
    "Cosine similarity measures vector similarity.",
    "Artificial Intelligence is transforming industries."
]
```

```python
response=client.models.embed_content(
    model="gemini-embedding-001",
    contents=chunks
)
```

```python
documents=[]
for chunks ,emb  in zip(chunks, response.embeddings):
  documents.append({
      "text":chunks,
      "embedding":emb.values
  })
```

```python
def cosine_similarity(a,b):
   a = np.array(a)
   b = np.array(b)

   return np.dot(a, b) / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )
```

```python
query = "What is Artificial Intelligence?"
```

```python
query_embedding = client.models.embed_content(
    model="gemini-embedding-001",
    contents=query
).embeddings[0].values
```

```python
scores = []
for doc in documents:
    score = cosine_similarity(
        query_embedding,
        doc["embedding"]
    )

    scores.append({
        "text": doc["text"],
        "score": score
    })

scores.sort(key=lambda x: x["score"], reverse=True)

print("Top 3 Results:\n")

for result in scores[:3]:
    print(f"Score: {result['score']:.4f}")
    print(result["text"])
    print("-" * 40)
```
